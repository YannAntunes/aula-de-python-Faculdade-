# Tratamento de valores duplicados com Python e Pandas



## 1. Importar as bibliotecas

Usaremos o **Pandas**, biblioteca bastante comum para manipulação e análise de dados. O `display()` ajuda a visualizar tabelas no Jupyter Notebook.


```python
import pandas as pd
from IPython.display import display

pd.set_option("display.max_rows", 60)
pd.set_option("display.max_columns", None)
```

## 2. Carregar o arquivo de dados

Coloque o arquivo `dados_duplicados.csv` na mesma pasta deste notebook. Se necessário, ajuste o caminho no `read_csv()`.


```python


df = pd.read_csv("../dados/dados_duplicados.csv")

print(f"Quantidade de linhas: {len(df)}")
print(f"Quantidade de colunas: {len(df.columns)}")
display(df.head(10))
```

## 3. O que é uma duplicata?

Uma linha duplicada é um registro que repete outro segundo os campos usados na comparação. Com `df.duplicated()`, o Pandas compara a linha com as linhas anteriores e marca como `True` as ocorrências repetidas.

Por padrão, a primeira ocorrência é marcada como `False` e as seguintes como `True`.


```python
df.duplicated()
```

## 4. Contar duplicatas completas

O método `sum()` soma os valores `True`, permitindo contar quantas linhas duplicadas existem.


```python
quantidade_duplicadas = df.duplicated().sum()
print(f"Duplicatas completas encontradas: {quantidade_duplicadas}")
```

## 5. Visualizar todas as linhas envolvidas em duplicações

Com `keep=False`, todas as ocorrências de um grupo duplicado são marcadas, inclusive a primeira. Isso ajuda a investigar antes de decidir o que remover.


```python
duplicatas = df[df.duplicated(keep=False)].sort_values(by=["id_aluno"])
display(duplicatas)
```

## 6. Remover linhas completamente duplicadas

`drop_duplicates()` retorna um DataFrame sem as linhas completamente duplicadas. Por padrão, mantém a primeira ocorrência. Aqui, guardamos o resultado em uma nova variável para preservar `df` original para comparação.


```python
df_sem_duplicatas = df.drop_duplicates()

print("Linhas antes:", len(df))
print("Linhas depois:", len(df_sem_duplicatas))
print("Duplicatas restantes:", df_sem_duplicatas.duplicated().sum())

display(df_sem_duplicatas.head())
```

## 7. Entender o parâmetro `keep`

- `keep="first"`: mantém a primeira ocorrência (padrão).
- `keep="last"`: mantém a última ocorrência.
- `keep=False`: remove todas as ocorrências que fazem parte de um grupo duplicado.

Faça testes nas células abaixo para comparar os resultados.


```python
# Mantém a primeira ocorrência
df.drop_duplicates(keep="first").head(10)
```

```python
# Mantém a última ocorrência
df.drop_duplicates(keep="last").head(10)
```

```python
# Remove todas as linhas que fazem parte de grupos duplicados
df.drop_duplicates(keep=False).head(10)
```

## 8. Procurar duplicatas em colunas específicas

Às vezes, queremos verificar repetição por uma chave, como `id_aluno`, em vez de comparar a linha inteira.

- `duplicated(subset=["id_aluno"])` identifica IDs repetidos.
- `drop_duplicates(subset=["id_aluno"], keep="last")` mantém o último registro de cada ID.

**Importante:** escolha `subset` de acordo com a regra do problema. Não use o nome como chave única se pessoas diferentes puderem ter o mesmo nome.


```python
duplicatas_por_id = df[df.duplicated(subset=["id_aluno"], keep=False)].sort_values("id_aluno")
display(duplicatas_por_id)
```

```python
df_um_registro_por_id = df.drop_duplicates(subset=["id_aluno"], keep="last")

print("Linhas originais:", len(df))
print("Linhas após manter um registro por ID:", len(df_um_registro_por_id))
```

## 9. Mesmo nome não significa necessariamente duplicata

Neste conjunto, Ana Souza aparece em registros diferentes com notas diferentes. São linhas diferentes; se removermos apenas pelo nome, poderemos apagar uma informação legítima.

A célula a seguir mostra esse caso.


```python
display(df[df["nome"] == "Ana Souza"])
```

```python
# Demonstração: manter somente a última linha por nome.
# Use apenas se essa for realmente a regra de negócio!
df_exemplo_por_nome = df.drop_duplicates(subset=["nome"], keep="last")
display(df_exemplo_por_nome[df_exemplo_por_nome["nome"] == "Ana Souza"])
```

## 10. Contar frequências de valores

`value_counts()` conta quantas vezes cada valor aparece em uma coluna. Ele é útil para encontrar valores repetidos, mas não remove linhas.


```python
print(df["nome"].value_counts().head(10))
```

## 11. Validar o resultado e salvar

Depois do tratamento, confira se ainda existem duplicatas segundo a regra escolhida. Salve em um arquivo novo para preservar os dados brutos.


```python
# Tratamento de exemplo: remover linhas completamente duplicadas
df_limpo = df.drop_duplicates()

print("Duplicatas completas antes:", df.duplicated().sum())
print("Duplicatas completas depois:", df_limpo.duplicated().sum())

df_limpo.to_csv("dados_alunos_limpos.csv", index=False, encoding="utf-8-sig")
print("Arquivo salvo: dados_alunos_limpos.csv")
```

## 12. Resumo 

| Método | Para que serve |
|---|---|
| `duplicated()` | Identifica linhas duplicadas |
| `duplicated().sum()` | Conta duplicatas completas |
| `duplicated(keep=False)` | Marca todas as ocorrências duplicadas |
| `drop_duplicates()` | Remove linhas duplicadas |
| `subset=["coluna"]` | Compara apenas colunas escolhidas |
| `keep="first"` | Mantém a primeira ocorrência |
| `keep="last"` | Mantém a última ocorrência |
| `keep=False` | Remove todas as ocorrências dos grupos duplicados |
| `value_counts()` | Conta a frequência de valores em uma coluna |
