```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv('../dados/dados_alunos.csv')
df
```

```python
df['idade'].plot.hist(bins=10, title='Histograma de Idades', figsize=(10, 6), color='blue', edgecolor='black') 
plt.xlabel('Idades')
plt.ylabel('Frequência')
plt.show()
```

```python
df['genero'].value_counts().plot.pie(title='Distribuição de Gênero', figsize=(8, 8), autopct='%1.1f%%', colors=['lightblue', 'lightpink'], wedgeprops={'width': 0.5})
plt.show()
```

```python
df['genero'].value_counts().plot.barh(title='Histograma de Gênero', figsize=(10, 6), color='red', edgecolor='black') 
plt.xlabel('quantidade')
plt.ylabel('gênero')
plt.xticks(rotation=0) 
plt.show()
```
