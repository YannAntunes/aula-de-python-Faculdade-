```python
# importando bibliotecas de analise de dados
import pandas as pd
```

```python
dados = {
    "Nome": ["Alice", "Bob", "Charlie", "David"],
    "Idade": [25, 30, 35, 40],
    "Altura": [1.65, 1.80, 1.75, 1.70],
    "Curso": ["Engenharia", "Medicina", "Direito", "Ciência da Computação"],
    "Escolaridade": ["Medio", "Superior", "Superior", "Medio"],
    "Aprovado": [True, False, True, True],
    "Data_Matricula": pd.to_datetime(["2021-01-15", "2020-09-10", "2022-03-05", "2021-11-20"])
}

df = pd.DataFrame(dados)
```

```python
# convertendo a coluna 'Data_Matricula' para o tipo datetime

df['Data_Matricula'] = pd.to_datetime(df['Data_Matricula'])
```

# Tipos de variaveis 
## Numericas 

- Discretas: assumem numeros inteiros
- Continuas: podem assumir numeros em qualquer intervalo de valor {ex: altura, peso} 

```python
print("Idade (variavel numerica Descreta)")
print(df['Idade'])
print("Altura (variavel numerica Continua)")
print(df['Altura'])
```

# Variaveis Categoricas 

- Nominais: nao tem ordem ou hieraquia {ex: curso}
- Ordinais: possuem ordem natural, ordem logica ou hieraquica {ex: escolaridade}

```python
print("Curso (variavel categorica nominal)")
print(df['Curso'])
print("Escolaridade (variavel categorica ordinal)")
print(df['Escolaridade'])
```

# Variaveis Binarias

sao variaveis com apenas dois valores possiveis.


```python
print("Aprovado (variavel categorica binaria)")
print(df['Aprovado'])
```

# Variavel Temporais 

Representam tempo ou sequencia de eventos.

```python
print("Data_Matricula (variavel de temporal)")
print(df['Data_Matricula'])
```
